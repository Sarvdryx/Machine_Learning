### Step 1. Import the necessary libraries:

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Tips/tips.csv). 

### Step 3. Assign it to a variable called tips

```python
url = 'https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Tips/tips.csv'
tips = pd.read_csv(url)
```

```python
tips
```

### Step 4. Delete the Unnamed 0 column

```python
del tips['Unnamed: 0']
```

### Step 5. Plot the total_bill column histogram

```python
tips['total_bill'].plot(kind='hist', bins=20, edgecolor='black', color='skyblue')

plt.title('Histogram of Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()

# Với kind='bar': Pandas sẽ cố gắng vẽ một cột (hoặc một cụm cột) cho từng dòng dữ liệu riêng lẻ trong bảng (ví dụ bảng tips có hơn 200 dòng, nghĩa là có hơn 200 nhãn giá trị chen chúc trên trục X). Do đó, chữ bị đè lên nhau thành một vệt đen đặc.

# Với kind='hist': Dữ liệu liên tục (như total_bill) sẽ được tự động gom nhóm lại thành một số ít các khoảng giá trị gọi là bins (thường chỉ khoảng từ 10 đến 20 cột/khoảng). Trục X lúc này chỉ hiển thị các khoảng đại diện đó nên rất thoáng và dễ nhìn.
```

### Step 6. Create a scatter plot presenting the relationship between total_bill and tip

```python
plt.scatter(
    x=tips['total_bill'],
    y=tips['tip']
)

plt.title('Relationship between total_bill and tip')
plt.xlabel('Total bill')
plt.ylabel('Tip')

plt.show()
```

### Step 7.  Create one image with the relationship of total_bill, tip and size.
#### Hint: It is just one function.

```python
sns.pairplot(tips)
plt.show()
```

### Step 8. Present the relationship between days and total_bill value

```python
sns.boxplot(x='day', y='total_bill', data=tips)
plt.show()
```

### Step 9. Create a scatter plot with the day as the y-axis and tip as the x-axis, differ the dots by sex

```python
sns.scatterplot(x='tip', y='day', hue='sex', data=tips)
plt.show()
```

### Step 10.  Create a box plot presenting the total_bill per day differetiation the time (Dinner or Lunch)

```python
sns.boxplot(x='day', y='total_bill', hue='time', data=tips)
plt.show()
```

### Step 11. Create two histograms of the tip value based for Dinner and Lunch. They must be side by side.

```python
sns.displot(data=tips, x='tip', col='time', kind='hist')
plt.show()
```

### Step 12. Create two scatterplots graphs, one for Male and another for Female, presenting the total_bill value and tip relationship, differing by smoker or no smoker
### They must be side by side.

```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='smoker',  # Phân biệt màu sắc theo nhóm hút thuốc hay không
    col='sex',  # Tách thành 2 biểu đồ nằm cạnh nhau dựa theo giới tính (Male và Female)
    kind='scatter',  # Chọn loại biểu đồ là scatter plot
)
plt.show()
```
