```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import matplotlib.pyplot as plt
```

```python
TRAIN_DIR = "/kaggle/input/datasets/utkarshsaxenadn/flower-classification-5-classes-roselilyetc/Flower Classification V2/V2/Testing Data"
VAL_DIR = "/kaggle/input/datasets/utkarshsaxenadn/flower-classification-5-classes-roselilyetc/Flower Classification V2/V2/Validation Data"
TEST_DIR = "/kaggle/input/datasets/utkarshsaxenadn/flower-classification-5-classes-roselilyetc/Flower Classification V2/V2/Testing Data"
```

```python
train_datagen = ImageDataGenerator(
    rescale = 1. / 255,
    shear_range = 0.2,
    zoom_range = 0.2,
    horizontal_flip = True
)
train_set = train_datagen.flow_from_directory(TRAIN_DIR, target_size = (224, 224), batch_size = 32, class_mode = 'categorical')

val_datagen = ImageDataGenerator ( rescale = 1. /255 )
val_set = val_datagen.flow_from_directory(VAL_DIR, target_size = (224, 224), batch_size = 32, class_mode = 'categorical')
```

```python
# 1. Khai báo mô hình gốc MobileNetV2 (đã bỏ phần classifier ở đỉnh)
base_model = tf.keras.applications.MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights='imagenet'
)

# Đóng đóng đóng băng các trọng số của base_model
base_model.trainable = False

# 2. Xây dựng mô hình mới
model = tf.keras.models.Sequential([
    base_model,
    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(10, activation='softmax')
])

# 3. Compile với Learning Rate phù hợp cho Transfer Learning
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# 4. Huấn luyện
history = model.fit(
    train_set,
    validation_data=val_set,
    epochs=20
)

# Sửa lỗi truy cập dict bằng dấu ngoặc vuông [] thay vì () và lỗi chính tả 'hisotry'
acc = history.history['accuracy']
val_acc = history.history['val_accuracy']
loss = history.history['loss']
val_loss = history.history['val_loss']

print("Train Accuracy:", acc)
print("Val Accuracy:", val_acc)

# ==========================================
# CÁCH LƯU MÔ HÌNH VÀO TỆP
# ==========================================

# Cách 1: Lưu toàn bộ mô hình (kiến trúc + trọng số) định dạng Keras (.keras)
model.save("flower_model.keras")
print("Đã lưu mô hình thành công tại /kaggle/working/flower_model.keras")
```

```python
test_datagen = ImageDataGenerator(rescale=1. / 255)

test_set = test_datagen.flow_from_directory(
    TEST_DIR,
    target_size=(224, 224),
    batch_size=32,
    class_mode='categorical',
    shuffle=False  # Đặt False để giữ đúng thứ tự ảnh khi dự đoán/đánh giá
)

# 1. Đường dẫn đến file mô hình đã lưu
model_path = "/kaggle/working/flower_model.keras"

# 2. Load lại mô hình
loaded_model = tf.keras.models.load_model(model_path)

# ==========================================
# 2. ĐÁNH GIÁ MÔ HÌNH TRÊN TẬP TEST
# ==========================================
test_loss, test_acc = model.evaluate(test_set)

print("-" * 40)
print(f"Test Loss    : {test_loss:.4f}")
print(f"Test Accuracy: {test_acc * 100:.2f}%")
print("-" * 40)
```
